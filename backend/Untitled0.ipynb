```python
import pandas as pd
train_url = "https://huggingface.co/datasets/sachin-iitd/DelhiPollDataset/resolve/main/train.csv"

train_df = pd.read_csv(train_url)

train_df.head()
```

```python
train_df["dateTime"] = pd.to_datetime(train_df["dateTime"])
```

```python
train_df = train_df.sort_values("dateTime")
```

```python
train_df[["dateTime", "pm2_5"]].head(10)
```

```python
train_df = train_df.reset_index(drop=True)
```

```python
train_df[["lat", "long"]].drop_duplicates().shape
```

```python
train_df[["lat", "long"]].drop_duplicates().head(10)
```

```python
duplicates = train_df.duplicated(
      subset=["lat", "long", "dateTime"],
          keep=False
)
duplicates.sum()
```

```python
train_df["future_pm2_5"] = (
      train_df.groupby(["lat", "long"])["pm2_5"].shift(-1)
      )
```

```python
train_df[["lat", "long", "dateTime", "pm2_5", "future_pm2_5"]].head(10)
```

```python
train_df["future_pm2_5"].isna().sum()
```

```python
train_df = train_df.dropna(subset=["future_pm2_5"])
```

```python
train_df.shape
```

```python
X = train_df.drop(columns=["future_pm2_5"])
y = train_df["future_pm2_5"]
```

```python
X.head()
```

```python
y.head()
```

```python
X.shape
```

```python
y.shape
```

```python
train_df["hour"] = train_df["dateTime"].dt.hour
```

```python
train_df["day_of_week"] = train_df["dateTime"].dt.dayofweek
```

```python
train_df[["dateTime", "hour", "day_of_week"]].head(10)
```

```python
train_df["month"] = train_df["dateTime"].dt.month
```

```python
train_df[["dateTime", "hour", "day_of_week", "month"]].head(10)
```

```python
X = train_df.drop(columns=["future_pm2_5", "dateTime"])
y = train_df["future_pm2_5"]
```

```python
split = int(len(train_df) * 0.8)

X_train = X.iloc[:split]
X_val = X.iloc[split:]

y_train = y.iloc[:split]
y_val = y.iloc[split:]
```

```python
X_train.shape, X_val.shape, y_train.shape, y_val.shape
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
model = RandomForestRegressor(
      n_estimators=100,
          random_state=42,
              n_jobs=-1
)
```

```python
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_val)
```

```python
predictions[:10]
```

```python
y_val.head(10).values
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_val, predictions)

print("MAE:", mae)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(mean_squared_error(y_val, predictions))

print("RMSE:", rmse)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_val, predictions)

print("R²:", r2)
```

```python
train_df["pm2_5_lag1"] = (
      train_df.groupby(["lat", "long"])["pm2_5"].shift(1)
)
```

```python
train_df[[
      "lat",
          "long",
              "dateTime",
                  "pm2_5",
                      "pm2_5_lag1"
                      ]].head(10)
```

```python
train_df["pm2_5_lag2"] = (
      train_df.groupby(["lat", "long"])["pm2_5"].shift(2)
      )

train_df["pm2_5_lag3"] = (
      train_df.groupby(["lat", "long"])["pm2_5"].shift(3)
      )
```

```python
train_df[[
      "dateTime",
          "pm2_5",
              "pm2_5_lag1",
                  "pm2_5_lag2",
                      "pm2_5_lag3"
                      ]].head(10)
```

```python
train_df = train_df.dropna(
      subset=["pm2_5_lag1", "pm2_5_lag2", "pm2_5_lag3"]
      )
```

```python
train_df.shape
```

```python
features = [
      "lat",
          "long",
              "pressure",
                  "temperature",
                      "humidity",
                          "pm1_0",
                              "pm2_5",
                                  "pm10",
                                      "hour",
                                          "day_of_week",
                                              "month",
                                                  "pm2_5_lag1",
                                                      "pm2_5_lag2",
                                                          "pm2_5_lag3"
                                                          ]

X = train_df[features]
y = train_df["future_pm2_5"]
```

```python
split = int(len(train_df) * 0.8)

X_train = X.iloc[:split]
X_val = X.iloc[split:]

y_train = y.iloc[:split]
y_val = y.iloc[split:]
```

```python
X_train.shape, X_val.shape, y_train.shape, y_val.shape
```

```python
model2 = RandomForestRegressor(
      n_estimators=100,
          random_state=42,
              n_jobs=-1
              )

model2.fit(X_train, y_train)
```

```python
predictions2 = model2.predict(X_val)
```

```python
predictions2[:10]
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae2 = mean_absolute_error(y_val, predictions2)
rmse2 = np.sqrt(mean_squared_error(y_val, predictions2))
r2_2 = r2_score(y_val, predictions2)

print("MAE:", mae2)
print("RMSE:", rmse2)
print("R²:", r2_2)
```

```python
import joblib

joblib.dump(model2, "airguard_random_forest.pkl")
```
